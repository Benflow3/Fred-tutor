```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
import os

list_of_class = ['Primary 1 Scheme of Work', 'Primary 2 Scheme of Work', 'Primary 3 Scheme of Work', 'Primary 4 Scheme of Work', 'Primary 5 Scheme of Work', 'Primary 6 Scheme of Work']
list_of_subject = ["English Studies", "Cultural & Creative Art", "Mathematics", "Civic Education", "Social Studies", "Security Education", "Yoruba Language","History", "Christian Religious Studies", "Basic Science & Technology", "Basic Science", "Physical & Health Education", "Information Technology"]
syllabus_url = "https://syllabus.ng/?srsltid=AfmBOoqCFLmxDgEovVVtUEmO8Ndy75MW_ttwNVuNFajPxy5ufWVYtz8#"

# Define the path to the raw dataset storage folder
output_folder = "Fred-tutor/raw_dataset_storage"
os.makedirs(output_folder, exist_ok=True)

try:
    req = requests.get(syllabus_url)
    req.raise_for_status() # Raise an exception for bad status codes (4xx or 5xx)

    soup = BeautifulSoup(req.text, "lxml")

    #first grab all the link
    for link in soup.find_all("a"):
        #check if the string of current link is in the list of class
        if link.text in list_of_class:
            class_name = link.text.replace(" ", "_") # Create a clean name for the class
            #then get the link of the class and load it
            curent_link = link.get("href")
            req = requests.get(curent_link)
            req.raise_for_status() # Raise an exception for bad status codes (4xx or 5xx)
            soup = BeautifulSoup(req.text, "lxml")
            #grab all the link
            for subject_link in soup.find_all("a"):
                subject_name_raw = subject_link.text[11:].strip()
                #then check if the string of subject link is present in list of subject
                if subject_name_raw in list_of_subject:
                    subject_name = subject_name_raw.replace(" ", "_").replace("&", "and") # Create a clean name for the subject
                    curent_subject_link = subject_link.get("href")
                    req = requests.get(curent_subject_link)
                    req.raise_for_status() # Raise an exception for bad status codes (4xx or 5xx)
                    soup = BeautifulSoup(req.text, "lxml")

                    # then scrape the table
                    tables = soup.find_all('table')
                    for i, table in enumerate(tables):
                        try:
                            df = pd.read_html(str(table))[0]
                            # Sanitize filename and include folder path
                            filename = os.path.join(output_folder, f"{class_name}_{subject_name}_table_{i+1}.csv")
                            df.to_csv(filename, index=False)
                            print(f"Saved {filename}")
                        except Exception as e:
                            print(f"Could not convert table {i+1} to DataFrame for {class_name} - {subject_name}: {e}")

            if link.text == 'Primary 6 Scheme of Work':
                break;
            # Remove this break statement if you want to process all classes
            # break;

except requests.exceptions.RequestException as e:
    print(f"Error fetching the URL: {e}")
except Exception as e:
    print(f"An error occurred: {e}")
```

##The code below is for scraping lesson note from classbasic site

###Algorithm to scrape the site



*   First go to the lesson note site
*   then grab all the week links and store it in list
*   then visit the first week link
*   then grab all the links
*   then move to the first link
*   check if its text contains subject and primary or  teaching and primary
*   if yes visit the link and scrape the body into csv and into lesson folder
*   do it for all the link then move to second week link
*   repeat the 4th step to the last week



```python
import os
import cloudscraper
from bs4 import BeautifulSoup

output_folder = "Fred-tutor/raw_lessonNote_storage"
os.makedirs(output_folder, exist_ok=True)

lesson_note_site = "https://classbasic.com/"

scrape_lesson_notes = cloudscraper.create_scraper()
request = scrape_lesson_notes.get(lesson_note_site)
soup = BeautifulSoup(request.text, "lxml")

all_week_link = [link.get('href') for link in soup.find_all('a') if link.text.startswith("Week")]

for week_link in all_week_link:
  request = scrape_lesson_notes.get(week_link)
  soup = BeautifulSoup(request.text, "lxml")
  #print(request.text) use for debuging
  div = soup.find('div', id='content')
  for subject_link in div.find_all('a'):
    #print(subject_link.text) use for debuging
    fulfilled = subject_link.text.find('Primary') != -1
    fulfilled = not(subject_link.text.find('textbooks') != -1 or subject_link.text.find('workbooks') != -1) and fulfilled
    print(fulfilled)
    if fulfilled:
      request = scrape_lesson_notes.get(subject_link.get('href'))
      soup = BeautifulSoup(request.text, "lxml")
      topic_name = subject_link.text.replace(" ", "_").replace("&", "and")
      filename = os.path.join(output_folder, f"{topic_name}.txt")
      body_content = soup.body.get_text(" ",strip=True)
      try:
          with open(filename, 'w') as file:
              file.write(body_content)
              print(f"Saved {filename}")
      except Exception as e:
          print(f"Could not save {filename}: {e}")
```
