```python
#the file of a particular subject must not be greater or lower than 3
#this cell will print out the files and number of copy
import os
import re

file_folder = '/content/Fred-tutor/raw_shcemeOfwork_storage'
list_of_file = os.listdir(file_folder)
remove_pattern = r'\_table\_[0-9]+\.csv$' # Modified regex pattern to remove '_table_number.csv'
new_list_of_file = [re.sub(remove_pattern,'',new_file) for new_file in list_of_file]
#print(new_list_of_file)
def group(files):
  collector ={}
  for file in files:
    if file in collector:
      collector[file] +=1
    else:
      collector[file] = 1
  return collector

duplicate_check = group(new_list_of_file)
for each in duplicate_check:
  print(each,duplicate_check[each])
```

##I want to turn this csv file to txt file for easier use in rag in this format below




*   subject title | class| term
*   week: number
* topic: topic title
* objectives : objecive




```python
#first create a folder for the final file
import os
import csv
from glob import glob

folder_name = '/content/Fred-tutor/final_shcemeOfwork_storage'
if not os.path.exists(folder_name):
  os.makedirs(folder_name)

#folder containng the csv files
csv_folder = '/content/Fred-tutor/raw_shcemeOfwork_storage'

#process each csv in the folder to look like the above format
for csv_path in glob(os.path.join(csv_folder,'*.csv')):
  with open(csv_path,'r') as csv_file:
    csv_reader = list(csv.reader(csv_file))
    class_name = None
    subject = None
    term = None
    header_row = None
    for i,row in enumerate(csv_reader):
      if 'CLASS' in row or 'Class' in row:
        row_index = row.index('CLASS') if 'CLASS' in row else row.index('Class')
        class_name = row[row_index+1]
      if 'SUBJECT' in row or 'Subject' in row:
        row_index = row.index('SUBJECT') if 'SUBJECT' in row else row.index('Subject')
        subject = row[row_index+1]
      if 'TERM' in row or 'Term' in row:
        row_index = row.index('TERM') if 'TERM' in row else row.index('Term')
        term = row[row_index+1]
      if 'WEEK' in row or 'OSE' in row or 'Week' in row or 'WEEKS' in row:
        header_row = i
        #print(f'Header row found at index {i}')
        break
    #print('after break')
    if header_row is None:
        print(f'No header row containing "WEEK" found in {csv_path}. Skipping.')
        continue

    headers = [h.strip().upper() for h in csv_reader[header_row]]
    #print(headers)
    # Check if required headers exist
    #required_headers = ['WEEK', 'TOPIC', 'OBJECTIVES','OSE']
    #if not all(header in required_headers for header in headers):
     #   print(f'Missing one or more required headers ({required_headers}) in {csv_path}. Skipping.')
      #  continue

    weeK_col = 0
    topic_col = 1
    objective_col = 2

    new_filename = os.path.join(folder_name,f'{subject}_{class_name}_{term}.txt')
    os.makedirs(os.path.dirname(new_filename),exist_ok=True)
    with open(new_filename,'w') as txt_file:
        txt_file.write(f'###{subject} | {class_name} | {term}\n')
        for row in csv_reader[header_row+1:]:
            week = row[weeK_col].strip()
            topic = row[topic_col].strip()
            objective = row[objective_col].strip()
            txt_file.write(f'Week: {week}\n')
            txt_file.write(f'Topic: {topic}\n')
            txt_file.write(f'Objectives: {objective}\n')
        print(f'{class_name} {subject} {term} completed\n')
```
